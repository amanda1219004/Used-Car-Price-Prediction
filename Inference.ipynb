# 1. Import Libraries

```python
import pandas as pd
import joblib
```

# 2. Load Model

```python
model = joblib.load(
    'gradient_boosting_tuned.joblib'
)
model
selected_features = joblib.load(
    'selected_features.joblib'
)
```

# 3. Buat Data Inference

```python
data_inf = pd.DataFrame({
    'name': ['Maruti Swift Dzire'],
    'year': [2018],
    'km_driven': [45000],
    'fuel': ['Diesel'],
    'seller_type': ['Individual'],
    'transmission': ['Manual'],
    'owner': ['First Owner'],
    'mileage(km/ltr/kg)': [22.0],
    'engine': [1248.0],
    'max_power': [74.0],
    'seats': [5.0]
})

data_inf
```

# 4. Feature Engineering

```python
REFERENCE_YEAR = 2020

data_inf['car_age'] = (
    REFERENCE_YEAR
    - data_inf['year']
)

data_inf['brand'] = (
    data_inf['name']
    .str.split()
    .str[0]
)

data_inf = data_inf.drop(
    columns=['name', 'year']
)

data_inf
```

```python
data_inf_final = data_inf[
    selected_features
]
data_inf_final
```

# 5. Predict

```python
prediction = model.predict(
    data_inf_final
)
print(
    f'Predicted Selling Price: '
    f'₹{prediction[0]:,.2f}'
)
```

#### Inference Conclusion

The trained **Gradient Boosting Regressor** successfully generates a predicted selling price based on the input vehicle characteristics.

This demonstrates that the saved model can be used to perform inference on new, unseen data and provide estimated used car selling prices.
